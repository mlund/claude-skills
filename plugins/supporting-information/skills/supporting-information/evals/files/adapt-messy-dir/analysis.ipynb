# Lysozyme in NaCl: ion distribution


```python
import os
import numpy as np
import matplotlib.pyplot as plt
```

```python
datadir = '/home/jane/sims/lyso-nacl/run'
traj = os.path.join(datadir, 'md.xtc')
```

```python
def rdf(r, nbins, rmax, volume):
    # radial distribution from pair distances
    hist, edges = np.histogram(r, bins=nbins, range=(0, rmax))
    shell = 4 / 3 * np.pi * (edges[1:] ** 3 - edges[:-1] ** 3)
    n = len(r)
    return 0.5 * (edges[1:] + edges[:-1]), hist / shell / (n / volume)
```

```python
r = np.loadtxt(os.path.join(datadir, 'distances.dat'))
x, g = rdf(r, 200, 3.0, 1000.0)
np.savetxt('/home/jane/sims/lyso-nacl/results/rdf.csv', np.c_[x, g], delimiter=',')
```

```python
plt.plot(x, g)
plt.savefig('fig2.png')
```
